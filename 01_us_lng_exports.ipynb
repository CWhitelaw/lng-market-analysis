```python
european_destinations = {
    "NUS-NFR": "France",
    "NUS-NGM": "Germany",
    "NUS-NNL": "Netherlands",
    "NUS-NSP": "Spain",
    "NUS-NBE": "Belgium",
    "NUS-NIT": "Italy",
    "NUS-NUK": "UK",
    "NUS-NPL": "Poland",
    "NUS-NPO": "Portugal",
    "NUS-NGR": "Greece",
    "NUS-NLH": "Lithuania",
}

all_dfs = []

for code, country in european_destinations.items():
    params = {
        "api_key": API_KEY,
        "frequency": "monthly",
        "data[0]": "value",
        "facets[duoarea][]": code,
        "facets[process][]": "EVE",
        "start": "2018-01",
        "sort[0][column]": "period",
        "sort[0][direction]": "asc",
        "length": 5000,
    }
    response = requests.get(url, params=params)
    data = response.json()
    if data["response"]["data"]:
        df = pd.DataFrame(data["response"]["data"])
        df["country"] = country
        all_dfs.append(df)

europe_df = pd.concat(all_dfs, ignore_index=True)

volume_df = europe_df[europe_df["units"] == "MMCF"].copy()
volume_df["period"] = pd.to_datetime(volume_df["period"])
volume_df["value"] = pd.to_numeric(volume_df["value"])

print(volume_df.shape)
print(volume_df["country"].unique())
print(volume_df.head())
```

```python
european_destinations = {
    "NUS-NFR": "France",
    "NUS-NGM": "Germany",
    "NUS-NNL": "Netherlands",
    "NUS-NSP": "Spain",
    "NUS-NBE": "Belgium",
    "NUS-NIT": "Italy",
    "NUS-NUK": "UK",
    "NUS-NPL": "Poland",
    "NUS-NPO": "Portugal",
    "NUS-NGR": "Greece",
    "NUS-NLH": "Lithuania",
}

all_dfs = []

for code, country in european_destinations.items():
    params = {
        "api_key": API_KEY,
        "frequency": "monthly",
        "data[0]": "value",
        "facets[duoarea][]": code,
        "facets[process][]": "EVE",
        "start": "2018-01",
        "sort[0][column]": "period",
        "sort[0][direction]": "asc",
        "length": 5000,
    }
    response = requests.get(url, params=params)
    data = response.json()
    if data["response"]["data"]:
        df = pd.DataFrame(data["response"]["data"])
        df["country"] = country
        all_dfs.append(df)

europe_df = pd.concat(all_dfs, ignore_index=True)

volume_df = europe_df[europe_df["units"] == "MMCF"].copy()
volume_df["period"] = pd.to_datetime(volume_df["period"])
volume_df["value"] = pd.to_numeric(volume_df["value"])

print(volume_df.shape)
print(volume_df["country"].unique())
print(volume_df.head())
```

```python
fig, ax = plt.subplots(figsize=(14, 7))

for country in volume_df["country"].unique():
    country_data = volume_df[volume_df["country"] == country]
    ax.plot(country_data["period"], country_data["value"], label=country, linewidth=1.5)

ax.axvline(pd.Timestamp("2022-02-24"), color="red", linestyle="--", linewidth=1.5, label="Ukraine Invasion")

ax.set_title("U.S. LNG Exports to Europe by Destination (2018–Present)", fontsize=14)
ax.set_xlabel("Date")
ax.set_ylabel("Volume (MMCF)")
ax.legend(loc="upper left", fontsize=8)
plt.tight_layout()
plt.savefig("us_lng_exports_europe.png", dpi=150)
plt.show()
```

```python

total_europe = volume_df.groupby("period")["value"].sum().reset_index()
total_europe.columns = ["period", "total_volume"]

fig, ax = plt.subplots(figsize=(14, 6))

ax.fill_between(total_europe["period"], total_europe["total_volume"], 
                alpha=0.4, color="steelblue", label="Total Europe")
ax.plot(total_europe["period"], total_europe["total_volume"], 
        color="steelblue", linewidth=1.5)

ax.axvline(pd.Timestamp("2022-02-24"), color="red", linestyle="--", 
           linewidth=1.5, label="Ukraine Invasion")

ax.set_title("Total U.S. LNG Exports to Europe (2018–Present)", fontsize=14)
ax.set_xlabel("Date")
ax.set_ylabel("Volume (MMCF)")
ax.legend()
plt.tight_layout()
plt.savefig("total_europe_lng.png", dpi=150)
plt.show()
```
